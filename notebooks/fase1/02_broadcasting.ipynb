1. Centralize e depois padronize (z-score) cada coluna de uma matriz (100, 4), sem loops.

```python
array = np.arange(400).reshape(100,4)
```

```python
mean = np.sum(array, axis=0) / 100
std = np.std(array, axis=0)
zscore = (array - mean) / std
zscore
```
