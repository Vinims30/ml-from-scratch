```python
import numpy as np

rng = np.random.default_rng(42)
```

```python
A = np.array([[1,2],[4,6]])
B = np.array([[1,2],[4,2],[0,0]])
print(A.shape)
print(B.shape)
```

```python
def distancia_2loops(A, B):
    D = np.zeros((A.shape[0], B.shape[0]))
    for i, linha in enumerate(A):
        for j, linhaj in enumerate(B):
            D[i,j] = np.sqrt(np.sum((linha - linhaj) **2))
    return D
```

```python
D = distancia_2loops(A, B)
print(D)
```

```python
def distancia_1loop(A,B):
    D = np.zeros((A.shape[0], B.shape[0]))
    for i, linha in enumerate(A):
        D[i] = np.sqrt(np.sum((linha - B) ** 2, axis=1))
    return D
```

```python
somaA = np.sum(A ** 2, axis=1)
somaA = somaA.reshape(-1,1)
somaA
```

```python
def distancia(A, B):
    D2 =  np.sum(A ** 2, axis=1).reshape(-1,1) + (np.sum(B ** 2,axis=1)  - 2 *(A @ B.T))
    D2 = np.maximum(D2,0)
    D = np.sqrt(D2)
    return D
```

```python
D = distancia(A, B)
print(D)
```
