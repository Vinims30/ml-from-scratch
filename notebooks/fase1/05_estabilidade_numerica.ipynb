```python
X = np.array([[1.43,-0.4,0.23],[1.43,-0.4,0.23]])

def softmax(x):
    exp = np.exp(x)
    softmax = exp / np.sum(exp, axis=1, keepdims=True)
    return softmax
```

```python
x_softmax = softmax(X)
x_softmax
```

```python
x_teste1 = np.array([[1000, 1001, 1002]])
x_teste2 = np.array([[-1000, -1001, -1002]])
```

```python
x_teste1soft = softmax(x_teste1)
x_teste1soft
```

```python
x_Teste2soft = softmax(x_teste2)
x_Teste2soft 
```

```python
def softmax_estavel(x):
    maior_linha = np.max(x,keepdims=True, axis=1)
    exp = np.exp(x - maior_linha)
    softmax = exp / np.sum(exp, axis=1, keepdims=True)

    return softmax
```

```python
x_softmax = softmax_estavel(X)
x_softmax
```

```python
x_teste1soft = softmax_estavel(x_teste1)
x_teste1soft
```

```python
x_Teste2soft = softmax_estavel(x_teste2)
x_Teste2soft 
```

```python
Z = np.array([
    [2.0,    -1.0,    0.5],   # caso comum
    [1.0,     1.0,    1.0],   # logits iguais
    [1000.0, 1001.0, 1002.0], # logits enormes (overflow na versão ingênua)
    [0.0,   -800.0,  -1.0],   # uma probabilidade vira 0 (underflow)
])

def log_softmax(Z):
    maior_linha = np.max(Z, keepdims=True,axis=1)
    exp = np.exp(Z - maior_linha)
    logsumexp = maior_linha + np.log(np.sum(exp, axis=1, keepdims=True))
    log_softmax = Z - logsumexp
    return log_softmax
```

```python
teste = log_softmax(Z)
teste
```

```python
esperado = np.array([[-0.241311, -3.241311, -1.741311],
                     [-1.098612, -1.098612, -1.098612],
                     [-2.407606, -1.407606, -0.407606],
                     [-0.313262, -800.313262, -1.313262]])
print(np.allclose(log_softmax(Z), esperado, atol=1e-6))
print(np.allclose(np.exp(log_softmax(Z)).sum(axis=1), 1)) 
```
